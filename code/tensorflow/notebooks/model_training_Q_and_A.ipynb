## Simple generator for loading data batches

```python
def simple_generator(labels,
                     directory,
                     indices,
                     batch_size = 32,
                     file_size = 8,
                     training_cycles = 1,
                     truncate_data = False,
                     shuffle_index = None):
    
    '''Simple generator function that loads data by the batches
       Since BERT embeddings are too large to fit into memory (~5TB),
           we must load them from disk so it is an expensive process
       Works with embeddings generated from: https://github.com/cbenge509/BERTVision/blob/master/notebooks/BERT_finetuning_Q_and_A_with_embedding_generation.ipynb
    '''

    
    if shuffle_index is None:
        np.random.shuffle(indices)
    else:
        indices = indices[shuffle_index]

    #Use a smaller fraction of the data by truncating randomized indices to specified number
    if truncate_data:
        indices = indices[:truncate_data]

    offset = 0
    splits = int(batch_size / file_size)
    
    while True:
        darray = np.zeros((batch_size,386,1024,25), dtype = np.float16)
        train_labels = [np.zeros(batch_size), np.zeros(batch_size)]
        
        if offset * splits > len(indices) - 1:
            offset = 0
            
        #Take a data slice
        data_slice = indices[offset * splits:(offset+1)*splits]
        #darray[:] = 0

        for i,idx in enumerate(data_slice):
            with h5py.File(directory + '%d.h5' %(idx), 'r') as f:
                data = np.array(f['hidden_state_activations'], dtype = np.float16)
                
            darray[i*file_size:(i+1)*file_size, :, :, :] = data
            
            train_labels[0][i*file_size:(i+1)*file_size] = labels[idx:idx+file_size].T[0]
            train_labels[1][i*file_size:(i+1)*file_size] = labels[idx:idx+file_size].T[1]

        if (offset + 1)*splits > len(indices) - 1:
            output = darray[:int(len(data_slice) * file_size)]
            labs = [train_labels[0][:int(len(data_slice) * file_size)],
                    train_labels[1][:int(len(data_slice) * file_size)]]
            offset = 0
            
        else:
            output = darray
            labs = train_labels
            offset += 1
        #print(output.shape)   
        yield output, labs
```

## Custom Layers

```python
class BertConcat(layers.Layer):
    '''Implementation of learned pooler reported by Tenney 2019
       Original paper: https://arxiv.org/abs/1905.05950
    '''
    def __init__(self, units = 1):
        super().__init__()
        
        #Will only work currently with units = 1
        self.units = 1
        
    def build(self, input_shape):
        self.w = self.add_weight(shape = (input_shape[-1],), trainable = True, initializer = 'random_normal', name = 'weights')
        self.t = self.add_weight(shape = (1), trainable = True, initializer = 'ones', name = 't')
    def call(self, inputs):
        w = tf.nn.softmax(self.w)
        return tf.reduce_sum(tf.multiply(inputs, w), axis = -1, keepdims = True) * self.t
```

```python
def gelu(x):
    """Gaussian Error Linear Unit.
    This is a smoother version of the RELU.
    Original paper: https://arxiv.org/abs/1606.08415
    Args:
      x: float Tensor to perform activation.
    Returns:
      `x` with the GELU activation applied.
    """
    cdf = 0.5 * (1.0 + tf.tanh(
        (np.sqrt(2 / np.pi) * (x + 0.044715 * tf.pow(x, 3)))))
    return x * cdf
    
class AdapterPooler(tf.keras.layers.Layer):
    '''Pooler inspired by Adapter BERT
       Original paper: https://arxiv.org/abs/1902.00751
       GitHub: https://github.com/google-research/adapter-bert
    '''
    
    def __init__(self, adapter_dim, init_scale = 1e-3, shared_weights = True):
        super().__init__()
        
        #Get the adapter size to be used
        self.adapter_dim = adapter_dim
        
        #Set initializer
        self.initializer = tf.keras.initializers.TruncatedNormal(stddev=init_scale)
        
        #Implement weight-sharing
        if shared_weights:
            self.pooler_layer = tf.keras.layers.TimeDistributed(
                tf.keras.layers.Dense(self.adapter_dim, kernel_initializer=self.initializer))
        else:
            self.pooler_layer = tf.keras.layers.LocallyConnected1D(self.adapter_dim, 1, 1, kernel_initializer=self.initializer)

    def call(self, inputs):
        '''Input shape expected to be (batch_size, 386, 1024, 25)
           Call reshapes tensor into (batch_size * 386, 25, 1024)
           Apply pooler_layer to input with gelu activation
        '''
        sequence_dim = inputs.shape[1]
        embedding_dim = inputs.shape[2]
        encoder_dim = inputs.shape[3]

        #Combine batch and sequence length dimension
        X = tf.reshape(inputs, [-1, embedding_dim, encoder_dim])

        #Move encoder_dim to axis = 1
        X = tf.transpose(X, (0, 2, 1))

        X = self.pooler_layer(X)
        X = gelu(X)

        #Regenerate shape
        X = tf.transpose(X, (0, 2, 1))
        X = tf.reshape(X, [-1, sequence_dim, self.adapter_dim, encoder_dim])

        return X
```

```python
def conv_block(inp, channels):
    '''Custom convolution blocks with 1xN kernels
       This preserves the dimensionality of axis 1
    '''
    kernel_init = tf.keras.initializers.glorot_uniform()
    bias_init = tf.keras.initializers.Constant(value = 0.2)
    l = layers.Conv2D(channels, (1,4), activation = 'relu', kernel_initializer = kernel_init, bias_initializer = bias_init)(inp)
    m = layers.Conv2D(channels, (1,8), activation = 'relu', kernel_initializer = kernel_init, bias_initializer = bias_init)(inp)
    n = layers.Conv2D(channels, (1,16), activation = 'relu', kernel_initializer = kernel_init, bias_initializer = bias_init)(inp)
    l = layers.MaxPool2D((1,4))(l)
    m = layers.MaxPool2D((1,8))(m)
    n = layers.MaxPool2D((1,16))(n)
    return tf.concat([l,m,n], axis = 2)
```

```python
def conv_block_nxn(inp, channels):
    '''Custom convolution blocks with NxN kernels with padding
       This preserves the dimensionality of axis 1
    '''
    kernel_init = tf.keras.initializers.glorot_uniform()
    bias_init = tf.keras.initializers.Constant(value = 0.2)
    l = layers.Conv2D(channels, (1,1), activation = 'relu', padding = 'same', kernel_initializer = kernel_init, bias_initializer = bias_init)(inp)
    m = layers.Conv2D(channels, (3,3), activation = 'relu', padding = 'same', kernel_initializer = kernel_init, bias_initializer = bias_init)(inp)
    n = layers.Conv2D(channels, (5,5), activation = 'relu', padding = 'same', kernel_initializer = kernel_init, bias_initializer = bias_init)(inp)
    
    #Maxpool only across the third axis
    m = layers.MaxPool2D((1,3))(m)
    n = layers.MaxPool2D((1,5))(n)
    return tf.concat([l,m,n], axis = 2)
```

```python
def conv_block_with_dilations(inp, channels):
    '''Custom convolution blocks with 1xN kernels with and without dilations
       This captures relationships between not just nearby embedding dimensions but also ones that are further away
       This preserves the dimensionality of axis 1
    '''
    kernel_init = tf.keras.initializers.glorot_uniform()
    bias_init = tf.keras.initializers.Constant(value = 0.2)

    l = layers.Conv2D(channels, (1,8), activation = 'relu', kernel_initializer = kernel_init, bias_initializer = bias_init)(inp)
    m = layers.Conv2D(channels, (1,16), activation = 'relu', kernel_initializer = kernel_init, bias_initializer = bias_init)(inp)
    n = layers.Conv2D(channels, (1,24), activation = 'relu', kernel_initializer = kernel_init, bias_initializer = bias_init)(inp)
    ld = layers.Conv2D(channels, (1,8), activation = 'relu', kernel_initializer = kernel_init, bias_initializer = bias_init, dilation_rate = [1, 2])(inp)
    md = layers.Conv2D(channels, (1,16), activation = 'relu', kernel_initializer = kernel_init, bias_initializer = bias_init, dilation_rate= [1, 2])(inp)
    nd = layers.Conv2D(channels, (1,24), activation = 'relu', kernel_initializer = kernel_init, bias_initializer = bias_init, dilation_rate = [1, 2])(inp)

    l = layers.MaxPool2D((1,8))(l)
    m = layers.MaxPool2D((1,16))(m)
    n = layers.MaxPool2D((1,24))(n)
    ld = layers.MaxPool2D((1,8))(ld)
    md = layers.MaxPool2D((1,16))(md)
    nd = layers.MaxPool2D((1,24))(nd)
    return tf.concat([l,m,n,ld,ld,nd], axis = 2)
```

# Models

### Learned and average pooling alone

```python
def learned_pooling():
    inp = layers.Input(shape = (386, 1024, 25), dtype = tf.float32)
    x = BertConcat()(inp)
    x = layers.Dense(2)(x)
    start, end = tf.split(x, 2, axis=-1)
    start = tf.squeeze(start, axis = -1)
    end = tf.squeeze(end, axis = -1)
    model = Model(inputs = inp, outputs = [start, end])
    return model
learned_pooling().summary()
```

```python
def average_pooling():
    inp = layers.Input(shape = (386, 1024, 25), dtype = tf.float32)
    x = tf.reduce_mean(inp, axis = -1, keepdims = True)
    x = layers.Dense(2)(x)
    start, end = tf.split(x, 2, axis=-1)
    start = tf.squeeze(start, axis = -1)
    end = tf.squeeze(end, axis = -1)
    model = Model(inputs = inp, outputs = [start, end])
    return model
average_pooling().summary()
```

### Learned pooling with CNN 

```python
def learned_pooling_CNN(dropout = 0.7):
    inp = layers.Input((386,1024,25))
    inp_seq = inp[:,:,:,-1]

    m = BertConcat()(inp)
    
    #Add 3 sets of convolution blocks
    m = conv_block(m, 32)
    m = conv_block(m, 64)
    m = conv_block(m, 128)
    m = tf.reshape(m, (-1, m.shape[1], m.shape[2] * m.shape[3]))
    m = tf.concat([m, inp_seq], axis = 2)
    if dropout:
        m = tf.keras.layers.Dropout(dropout)(m)
    m = layers.Dense(2)(m)

    start, end = tf.split(m, 2, axis=-1)

    start = tf.squeeze(start, axis = -1)
    end = tf.squeeze(end, axis = -1)

    model = Model(inputs=inp, outputs=[start, end], name = 'learned_pooling_cnn')
    return model

learned_pooling_CNN().summary()
```

### Adapter models

```python
def get_adapter_model(adapter_size = 386,
                      shared_weights = True,
                      add_skip = False):
    
    inp = layers.Input((386,1024,25))
    X = AdapterPooler(adapter_size, shared_weights=shared_weights)(inp)
    X = tf.reshape(X, (-1, X.shape[1], X.shape[2] * X.shape[3]))
    
    if add_skip:
        inp_seq = inp[:,:,:,-1]
        X = tf.concat([X, inp_seq], axis = 2)

    X = layers.Dense(2)(X)

    start, end = tf.split(X, 2, axis=-1)

    start = tf.squeeze(start, axis = -1)
    end = tf.squeeze(end, axis = -1)
    model = Model(inputs = inp, outputs = (start, end))
    return model

model_adapter_skip_shared_386 = get_adapter_model(386, shared_weights = True, add_skip = True)
model_adapter_skip_shared_386.summary()
```

### Adapter with 386 Pooling

```python
def get_adapter_with_pooling(pooling_first = True,
                             adapter_size = 386,
                             shared_weights = True,
                             add_skip = True):
    
    inp = layers.Input((386,1024,25))
    
    if pooling_first:
        X = BertConcat()(inp)
        X = tf.expand_dims(X, axis = -1)
        X = AdapterPooler(adapter_size, shared_weights = shared_weights)(X)
    else:
        X = AdapterPooler(adapter_size, shared_weights = shared_weights)(inp)
        X = BertConcat()(X)
        
    X = tf.reshape(X, (-1, X.shape[1], X.shape[2] * X.shape[3]))
    
    if add_skip:
        inp_seq = inp[:,:,:,-1]
        X = tf.concat([X, inp_seq], axis = 2)

    X = layers.Dense(2)(X)

    start, end = tf.split(X, 2, axis=-1)

    start = tf.squeeze(start, axis = -1)
    end = tf.squeeze(end, axis = -1)
    model = Model(inputs = inp, outputs = (start, end), name = 'adapter_tenney')
    return model

model_adapter_shared_skip_pooling_first = get_adapter_with_pooling()
model_adapter_shared_skip_pooling_first.summary()
```

### 1xN CNN with Adapter and Tenney

```python
#adapter + cnn + tenney
def get_base_cnn_model(dropout = 0.7):
    inp = layers.Input((386,1024,25))
    inp_seq = inp[:,:,:,-1]
    X = BertConcat()(inp)
    X = tf.expand_dims(X, axis = -1)
    X = AdapterPooler(386, shared_weights = False)(X)
    
    m = conv_block(X, 32)
    m = conv_block(m, 64)
    m = conv_block(m, 128)
    m = tf.reshape(m, (-1, m.shape[1], m.shape[2] * m.shape[3]))
    m = tf.concat([m, inp_seq], axis = 2)
    m = tf.keras.layers.Dropout(dropout)(m)
    m = layers.Dense(2)(m)

    start, end = tf.split(m, 2, axis=-1)

    start = tf.squeeze(start, axis = -1)
    end = tf.squeeze(end, axis = -1)

    model = Model(inputs=inp, outputs=[start, end])

    return model
get_base_cnn_model().summary()
```

### NxN CNN with Adapter and Learned Pooling

```python
def get_base_cnn_model_nxn(dropout = 0.7):
    inp = layers.Input((386,1024,25))
    inp_seq = inp[:,:,:,-1]
    X = BertConcat()(inp)
    X = tf.expand_dims(X, axis = -1)
    X = AdapterPooler(386, shared_weights = False)(X)

    m = conv_block_nxn(X, 32)
    m = conv_block_nxn(m, 64)
    m = conv_block_nxn(m, 128)
    m = tf.reshape(m, (-1, m.shape[1], m.shape[2] * m.shape[3]))
    m = tf.concat([m, inp_seq], axis = 2)
    m = tf.keras.layers.Dropout(dropout)(m)
    m = layers.Dense(2)(m)

    start, end = tf.split(m, 2, axis=-1)

    start = tf.squeeze(start, axis = -1)
    end = tf.squeeze(end, axis = -1, name = 'model_2_end_loss')

    model = Model(inputs=inp, outputs=[start, end])

    return model
get_base_cnn_model_nxn().summary()
```

### Learned pooling, Adapter, and Xception modified

```python
def get_xception_abbreviated():
    '''
    Modifier xception architecture to preserve the sequence length dimension
    Original paper: https://arxiv.org/abs/1610.02357
    '''

    input_img = layers.Input((386,1024,25), dtype = tf.float32)

    # pull the last channel layer for residual connection layer
    inp_seq = input_img[:,:,:,-1]
    #inp_seq = tf.squeeze(inp_seq, axis = 1)
    
    x = BertConcat()(input_img)
    x = tf.expand_dims(x, axis = -1)
    x = AdapterPooler(386, shared_weights = False)(x)
    
    # Block 1
    x = layers.Conv2D(64, (1, 3), strides=(1, 3), use_bias=False) (x)
    x = layers.BatchNormalization()(x)
    x = layers.Activation('relu')(x)
    x = layers.Conv2D(128, (1, 3), use_bias=False)(x)
    x = layers.BatchNormalization()(x)
    x = layers.Activation('relu')(x)

    residual = layers.Conv2D(512, (1, 1), strides=(1, 2), padding='same', use_bias=False)(x)
    residual = layers.BatchNormalization()(residual)

    # Block 2
    x = layers.SeparableConv2D(256, (1, 3), padding='same', use_bias=False)(x)
    x = layers.BatchNormalization()(x)
    x = layers.Activation('relu')(x)
    x = layers.SeparableConv2D(512, (1, 3), padding='same', use_bias=False)(x)
    x = layers.BatchNormalization()(x)

    # Block 2 Pool
    x = layers.AveragePooling2D((1, 3), strides=(1, 2), padding='same')(x)
    x = layers.add([x, residual])

    # Fully Connected Layer
    x = layers.AveragePooling2D((1, x.shape[2]))(x)
    x = tf.squeeze(x, axis = 2)

    # add the skip level residual back to the last CLS token
    x = layers.concatenate([x, inp_seq], axis = 2)

    x = layers.Dense(2, dtype = tf.float32, name = 'dense_2_final') (x)
    
    start, end = tf.split(x, 2, axis=-1)
    start = tf.squeeze(start, axis = -1)
    end = tf.squeeze(end, axis = -1)
    model = Model(inputs = input_img, outputs = [start, end])
    
    return model
get_xception_abbreviated().summary()
```

### Model compilation

All models are compiled as a bi-headed model, the first representing span start position and the second representing span end position. No activation is applied as the heads come directly from splitting a tensor.

```python
def compile_model(model):
    loss = tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True)
    
    #monitor accuracy during the training process
    model.compile(loss = [loss, loss],
                  optimizer='adam',
                  metrics = ['accuracy'])
```

### Example training procedure

```python
#Start generator with training labels, pointing to data directory with embeddings
data_dir = '../data/train_bert_1_epoch_fine_tuned_full386/'
gen = simple_generator(labels,
                 data_dir,   #data directory generated from https://github.com/cbenge509/BERTVision/blob/master/notebooks/BERT_finetuning_Q_and_A_with_embedding_generation.ipynb
                 batch_size = 16,
                 indices = list(range(len(labels)//8)), # since batch sizes of files is 8
                 file_size = 8,
                 training_cycles = 1,
                 truncate_data = None)
```

```python
#Gather a list of models to fit
#Since fitting is typically faster than data load, it is beneficial to many models at once
model_list = [model_adapter_skip_shared_386, 
              model_adapter_skip_independent_386,
              model_adapter_skip_shared_64,
              model_adapter_skip_independent_64]
```

```python
i = 0
max_batches = 8248 #Can be any number; this is pre-calculated based on the amount of training data used; 8248 goes through entire dataset at batch size of 16
for batch in gen:
    #Read in the batch of data from generator
    X = batch[0]
    Y = batch[1]
    
    for model_current in models_list:
        #Fit the generated dataset once
        model_current.fit(X, Y, epochs = 1)
    
    #increment counter
    i += 1
    
    #When the number of 
    if i == max_batches: #4 batches; can save each quarter
        break
```
